```python
#Step 2.1  Convert MCP tools to LangChain tools

from mcp.server.fastmcp import FastMCP
import threading

# Create MCP server
mcp = FastMCP("Math")

@mcp.tool()
def add(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b

@mcp.tool()
def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a * b

# Run server in background thread using streamable-http (notebook-friendly)
def run_server():
    mcp.run(transport="streamable-http")  # will listen on localhost (default)

threading.Thread(target=run_server, daemon=True).start()

print("MCP server started (streamable-http).")
```

```python
#Step 2.2 Connect to MCP server and list tools using the MultiServerMCPClient adapter
from langchain_mcp_adapters.client import MultiServerMCPClient

async def step2_connect_and_list():
    client = MultiServerMCPClient(
        {
            "math": {
                "transport": "http",
                "url": "http://127.0.0.1:8000/mcp",
            }
        }
    )

    tools = await client.get_tools()
    print("=== TOOLS ===")
    for t in tools:
        print(f"- {t.name}: {t.description}")

await step2_connect_and_list()
```

```python
#Step 3 Load MCP tools into a LangChain agent and test
from langchain_mcp_adapters.client import MultiServerMCPClient

async def step3_load_tools_and_test():
    client = MultiServerMCPClient(
        {
            "math": {
                "transport": "http",
                "url": "http://127.0.0.1:8000/mcp",
            }
        }
    )

    tools = await client.get_tools()

    print("=== LANGCHAIN TOOLS (from get_tools) ===")
    for t in tools:
        print(f"- {t.name} | type: {type(t)} | args_schema: {getattr(t, 'args_schema', None)}")

    # quick invocation test
    add_tool = next(t for t in tools if t.name == "add")
    out = await add_tool.ainvoke({"a": 2, "b": 3})
    print("\nTest: add(2,3) ->", out)

    return tools

lc_tools = await step3_load_tools_and_test()
```

```python
# Step 4 — Create Agent with MCP Tools (modern API)

from langchain_openai import ChatOpenAI
from langchain.agents import create_agent

# 1) LLM
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# 2) Create agent (v1 API)
agent = create_agent(
    model=llm,
    tools=lc_tools,
    system_prompt=(
        "You are a careful math assistant. "
        "Use the available tools for arithmetic. "
        "Return only the final numeric answer."
    ),
)

# 3) Test addition
res1 = await agent.ainvoke(
    {"messages": [("user", "What is 12 + 30?")]}
)
print("FINAL 1:", res1["messages"][-1].content)

# 4) Test multiplication
res2 = await agent.ainvoke(
    {"messages": [("user", "What is 7 times 8?")]}
)
print("FINAL 2:", res2["messages"][-1].content)
```

```python
# Step 5.1 Add resource to MCP server
@mcp.resource("info://rules")
def rules() -> str:
    """Read-only instructions for this server."""
    return (
        "Math MCP rules:\n"
        "- Use add(a,b) for addition.\n"
        "- Use multiply(a,b) for multiplication.\n"
        "- Return only the final number."
    )
```

```python
# Step 5.2. List resources and read resource content
from langchain_mcp_adapters.client import MultiServerMCPClient

async def step5_list_and_read():
    client = MultiServerMCPClient(
        {"math": {"transport": "http", "url": "http://127.0.0.1:8000/mcp"}}
    )

    resources = await client.get_resources("math")
    print("=== RESOURCES ===")
    for r in resources:
        print("-", r.metadata.get("uri"), "|", r.mimetype)

    text = await client.read_resource("math", "info://rules")
    print("\n=== CONTENT (info://rules) ===")
    print(text)

await step5_list_and_read()

from mcp import ClientSession
from mcp.client.streamable_http import streamablehttp_client

async def step5_read_resource_mcp_sdk():
    url = "http://127.0.0.1:8000/mcp"

    async with streamablehttp_client(url) as (read, write, _):
        async with ClientSession(read, write) as session:
            await session.initialize()

            # 1) List resources (proof)
            res_list = await session.list_resources()
            print("=== MCP list_resources() ===")
            for r in res_list.resources:
                print("-", r.uri, "|", r.mimeType)

            # 2) Read the resource
            res = await session.read_resource("info://rules")
            print("\n=== MCP read_resource(info://rules) ===")
            # read_resource returns a structured response; print text content if present
            for c in res.contents:
                # content can be TextResourceContents or BlobResourceContents
                text = getattr(c, "text", None)
                if text is not None:
                    print(text)
                else:
                    # fallback (e.g., blob)
                    print(c)

    return res

_ = await step5_read_resource_mcp_sdk()
```

```python
# Step 5.3 Use resource as context in agent
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from mcp import ClientSession
from mcp.client.streamable_http import streamablehttp_client

async def step5_use_resource_in_agent():
    # 1) Read resource text via MCP SDK
    url = "http://127.0.0.1:8000/mcp"
    async with streamablehttp_client(url) as (read, write, _):
        async with ClientSession(read, write) as session:
            await session.initialize()
            res = await session.read_resource("info://rules")
            rules_text = "\n".join(
                getattr(c, "text", str(c)) for c in res.contents
            )

    print("=== RULES TEXT (from resource) ===")
    print(rules_text)

    # 2) Build agent with the resource injected into system prompt
    llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
    agent = create_agent(
        model=llm,
        tools=lc_tools,
        system_prompt=(
            "You are a careful math assistant.\n\n"
            "Context (from MCP resource info://rules):\n"
            f"{rules_text}\n\n"
            "Follow the rules above."
        ),
    )

    # 3) Test: should still call tools and return only the number
    out = await agent.ainvoke({"messages": [("user", "Compute (9 + 11) * 2. Return only the number.")]})
    print("\nFINAL:", out["messages"][-1].content)

await step5_use_resource_in_agent()
```

In my installed langchain-mcp-adapters==0.2.1, MultiServerMCPClient supports get_resources() for listing but does not implement get_langchain_resources() or read_resource() (AttributeError). Therefore, resource reading was performed via the official MCP Python SDK (ClientSession.list_resources() / ClientSession.read_resource()), which fulfills the same Step 5 requirements.

```python
#To fulfull the handout-compatible function names for Steps 5.2 and 5.3, we can define two helper functions that wrap the MCP SDK calls. These functions will be named `get_langchain_resources()` and `read_resource(uri: str)`, matching the handout instructions.
from langchain_mcp_adapters.client import MultiServerMCPClient
from mcp import ClientSession
from mcp.client.streamable_http import streamable_http_client

SERVER_NAME = "math"
SERVER_URL = "http://127.0.0.1:8000/mcp"

async def get_langchain_resources():
    """Handout-compatible name: list resources (uses adapter's get_resources)."""
    client = MultiServerMCPClient({SERVER_NAME: {"transport": "http", "url": SERVER_URL}})
    return await client.get_resources(SERVER_NAME)

async def read_resource(uri: str):
    """Handout-compatible name: read resource content (uses MCP SDK)."""
    async with streamable_http_client(SERVER_URL) as (read, write, _):
        async with ClientSession(read, write) as session:
            await session.initialize()
            res = await session.read_resource(uri)
            return "\n".join(getattr(c, "text", str(c)) for c in res.contents)

# --- Demo (matches Step 5 wording) ---
resources = await get_langchain_resources()
print("=== get_langchain_resources() ===")
for r in resources:
    print("-", r.metadata.get("uri"), "|", r.mimetype)

rules_text = await read_resource("info://rules")
print("\n=== read_resource(info://rules) ===")
print(rules_text)
```

```python
#Step 6 Complete MCP enabled agent (MultiServer Client, tools, resources and real tests)
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain_mcp_adapters.client import MultiServerMCPClient

from mcp import ClientSession
from mcp.client.streamable_http import streamablehttp_client

SERVERS = {
    "math": {
        "transport": "http",
        "url": "http://127.0.0.1:8000/mcp",
        "resources": ["info://rules"],
    }
}

async def load_tools_from_servers(servers: dict):
    client = MultiServerMCPClient(
        {name: {"transport": cfg["transport"], "url": cfg["url"]} for name, cfg in servers.items()}
    )
    return await client.get_tools()

# 👇 THIS IS WHERE THE NEW FUNCTION GOES
async def read_resources_via_mcp_sdk(server_url: str, uris: list[str]) -> dict[str, str]:
    out = {}
    async with streamablehttp_client(server_url) as (read, write, _):
        async with ClientSession(read, write) as session:
            await session.initialize()

            res_list = await session.list_resources()
            available = [str(r.uri) for r in res_list.resources]
            print("Available resources on server:", available)

            for uri in uris:
                print("Reading:", uri)
                res = await session.read_resource(uri)
                text = "\n".join(getattr(c, "text", str(c)) for c in res.contents)
                out[uri] = text

    return out

async def build_complete_mcp_agent(servers: dict):
    tools = await load_tools_from_servers(servers)

    resources_text = {}
    for name, cfg in servers.items():
        server_url = cfg["url"]
        uris = cfg.get("resources", [])
        if uris:
            resources_text[name] = await read_resources_via_mcp_sdk(server_url, uris)
        else:
            resources_text[name] = {}

    context_blocks = []
    for server_name, uri_map in resources_text.items():
        for uri, text in uri_map.items():
            context_blocks.append(f"[MCP resource {server_name} :: {uri}]\n{text}")

    system_prompt = (
        "You are a helpful assistant with access to MCP tools.\n"
        "Use tools when needed.\n"
        "Follow MCP resource rules if provided.\n\n"
        + ("\n\n".join(context_blocks) if context_blocks else "")
    )

    llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
    agent = create_agent(model=llm, tools=tools, system_prompt=system_prompt)

    return agent, tools, resources_text

# ---- Run Step 6 ----
agent, tools, resources_text = await build_complete_mcp_agent(SERVERS)

print("Loaded tools:", [t.name for t in tools])
print("Loaded resources:", resources_text)
```
